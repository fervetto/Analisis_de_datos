#Analisis de Datos

## 1.- ETL (Extracción, Transformación y Carga)


### 1.1.- Extracción de datos

```python
import pandas as pd
from google.colab import drive
import chardet

result = chardet.detect(open('/content/drive/MyDrive/ventas.csv', 'rb').read())
print(result)
drive.mount('/content/drive', force_remount=True)
ruta_archivo = '/content/drive/MyDrive/ventas.csv'
df = pd.read_csv(ruta_archivo, encoding='ISO-8859-1', sep=';', decimal=',')
try:
  df.head()
  print('la Extracción fue correcta')
except FileNotFoundError:
  print(f'Error en la ruta del archivo: {ruta_archivo}')
```

### 1.2.- Transformación

```python
#Muestra información detallada de los tipos de datos en las columnas
#Renombrar las columnas

df.rename(columns={'precio de venta': 'precio',
                  'Tienda' : 'tienda',
                 'Marca' : 'marca',
                 'Tipo' : 'tipo',
                 'Gï¿½nero' : 'genero',
                 'Talla' : 'talla',
                 'Color' : 'color',
                 'Categoria' : 'categoria',
                 'Fecha' : 'fecha',
                 'Hora' : 'hora'}, inplace=True)

df['precio'] = df['precio'].str.replace(',', '.').astype(float)

```

```python



# Convertir la columna genero en tipo string
df['genero'] = df['genero'].astype(str)

# Reemplazar los valores en la columna genero


df['genero']=df['genero'].str.replace('Masculino', 'M')
df['genero']=df['genero'].str.replace('Femenino', 'F')
df['genero']=df['genero'].str.replace('Usinex', 'U')
print(set(df['genero']))

# Crear una columna combinando fecha y hora y pasarla al tipo de dato datatime
df['fecha_hora'] = pd.to_datetime(df['fecha'] + ' ' + df['hora'])
# Eliminar las columnas fecha y hora
df.drop(['fecha', 'hora'], axis=1, inplace=True)

# Tomamos el valor de la izquierda, en la columna talla en el caso de existir guiones -
#df['talla'] = df['talla'].str.split('-').str[0]

# Convertir a tipo numerico entero los datos en la columna talla
#df['talla'] = pd.to_numeric(df['talla'], errors='coerce').astype('Int64')

# Eliminar filas con talla nula o no numericas
#df = df.dropna(subset=['talla'])

print(df.info())
print(df.head())
```

### 1.3.- Carga

```python
ruta_salida = '/content/drive/MyDrive/ventas_transformado.csv'
df.to_csv(ruta_salida, index=False, sep=',')
print(f'El archivo se ha guardado en: {ruta_salida}')
```

## 2.- EDA


```python
# Cuanto facturó cada tienda
df = pd.read_csv('/content/drive/MyDrive/ventas_transformado.csv')
facturacion_ordenada = df.groupby(['tienda', 'marca'])['precio'].sum().reset_index()
top = facturacion_ordenada.sort_values(by='precio', ascending=False)
print(top)
```

```python
# Filtrado condicional -- Boolean indexing

# Definir la condición: categoría es pantalon & color es negro
condicion = (df['categoria'] == 'Pantalon') & (df['color'] == 'Negro')

# Aplicar la condición al DataFrame
df_filtrado = df[condicion]

# Cantidad de pantalones negros vendidos y monto total de los mismos
cantidad_vendida = df_filtrado['precio'].count()
monto_total = df_filtrado['precio'].sum()

print(f'Cantidad de pantalones negros vendidos: {cantidad_vendida} y el monto total es de: {monto_total}')
```
